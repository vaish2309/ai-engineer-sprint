Build a tiny vector comparison program

Import NumPy as np.



Create an array called candidate_a containing [2, 4, 6].



Create an array called candidate_b containing [1, 3, 5].



Print the sum of the two arrays.



Print the dot product of the two arrays.



Print the average of candidate_a.

Cosine similarity. It measures how closely two vectors point in the same direction.
- Similar direction → cosine similarity near 1
- Perpendicular directions → 0
- Opposite directions → near -1

```python
import numpy as np
candidate_a=np.array([2,4,6])
candidate_b=np.array([1,3,5])
sum_array=candidate_a+candidate_b
dot_array=np.dot(candidate_a,candidate_b)
print("Sum of two arrays: ",sum_array)
print("Dot product: ",dot_array)
print("Average of candidate_a :",np.mean(candidate_a))
length_a=np.linalg.norm(candidate_a)
print(length_a)
length_b=np.linalg.norm(candidate_b)
cosine_similarity=dot_array/(length_a*length_b)
print("Cosine similarity:",round(cosine_similarity,2))
```

Step 1 — Generate your first embedding
An embedding model converts text into a list of numbers. Unlike the Gemini API call you used to generate a written answer, this API call returns a numerical representation of the text.

Calculate cosine similarity
Now that we have the three vectors, we'll measure how similar their directions are.

```python
from google import genai
from google.colab import userdata
import numpy as np
api_key=userdata.get("Secret")
client=genai.Client(api_key=api_key)
print("Setup successful!")
```

What the code does:
- embed_content(...) sends your text to the embedding model.
- response.embeddings[0].values extracts the returned numerical vector.
- np.array(...) converts it into a NumPy array.
- shape and size tell you its structure and length.

```python
response = client.models.embed_content(
    model="gemini-embedding-001",
    contents="How do I reset my password?"
)

embedding = np.array(response.embeddings[0].values)

print("First 10 values:", embedding[:10])
print("Shape:", embedding.shape)
print("Number of values:", embedding.size)
```

```python
sentences = [
    "How do I reset my password?",
    "I forgot my login credentials.",
    "How do plants grow?"
]

embeddings=[]
for sentence in sentences:
  response = client.models.embed_content(
    model="gemini-embedding-001",
    contents=sentence
)
  embedding = np.array(response.embeddings[0].values)
  embeddings.append(embedding)
  print("Sentence:", sentence)
  print("Shape:", embedding.shape)
  print("-" * 40)

embeddings = np.array(embeddings)

print("All embeddings shape:", embeddings.shape)

def cosine_similarity(a,b):
  return np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))

print("Cosine of passowrd and login :",cosine_similarity(embeddings[0],embeddings[1]))
print("Cosine of passowrd and plants :",cosine_similarity(embeddings[0],embeddings[2]))
```

this compares the first sentence against all three sentences, including itself.

```python
query_index = 0
similarities = []

for i in range(len(embeddings)):
    score = cosine_similarity(
        embeddings[query_index],
        embeddings[i]
    )
    similarities.append(score)

for i, score in enumerate(similarities):
    print(f"{sentences[i]} -> {score:.4f}")
```
