Running the API Call

```python
from google import genai
from google.colab import userdata

# Read the API key from Colab Secrets
api_key = userdata.get("Secret")

# Create the Gemini client
client = genai.Client(api_key=api_key)

# Send a prompt to Gemini
response = client.interactions.create(
    model="gemini-3.8-flash",
    input="Explain what Python is in three simple sentences."
)

# Display Gemini's answer
print(response.output_text)
```

```python
numbers = [10, 0, 5]

for number in numbers:
    try:
        result = 100 / number
        print("Result:", result)
    except ZeroDivisionError:
        print("Cannot divide by zero")

print("Finished!")
```

```python
questions=["What is AI?","What is Python?","What is RAG?"]
for question in questions:
  print(question)
  try:
    response=client.interactions.create(
        model="gemini-3.8-flash",
      input=question)
    print("Answer:", response.output_text)
  except Exception as e:
    print("Request failed",e)

print("All questions processed!")
```
