Write a program that:
1. Defines a function called check_score that accepts one score.
2. If the score is at least 0.90, prints "Pass".
3. Otherwise, prints "Needs improvement".
4. Uses a for loop to call check_score for every score in the list.

scores = [0.95, 0.72, 0.88, 0.97]

```python
def check_score(score,pass_count):
  if score>=0.90:
    print("Pass")
    pass_count=pass_count+1
  else:
    print("Needs improvement")
  return pass_count

scores=[0.95, 0.60, 0.82, 0.91, 0.74]
pass_count=0
for score in scores:
  pass_count=check_score(score,pass_count)

print(pass_count)
```
